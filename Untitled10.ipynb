```python
from fastapi import FastAPI
import httpx, csv, io

app = FastAPI()
URL = "https://raw.githubusercontent.com/your-username/my-dataset/main/data.csv"

@app.get("/data")
async def get_data():
    async with httpx.AsyncClient() as client:
        r = await client.get(URL)
        r.raise_for_status()

    csv_text = r.text
    reader = csv.DictReader(io.StringIO(csv_text))
    rows = list(reader)
    return {"status": "success", "data": rows}
```
