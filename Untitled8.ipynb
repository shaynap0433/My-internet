```python
import os
import pandas as pd
import folium
from folium.plugins import FastMarkerCluster

# ==========================================
# 1. LOAD DATA & INITIAL CLEANING
# ==========================================
def load_and_clean_data(file_path):
    """
    Imports the IPEDS dataset, filters for required columns,
    and handles missing coordinates.
    """
    print("⌓ Loading dataset...")
    # Read CSV (IPEDS files often use ISO-8859-1 encoding)
    try:
        df = pd.read_csv(file_path, encoding='ISO-8859-1')
    except FileNotFoundError:
        print(f"❌ Error: The file '{file_path}' was not found. Please check your path.")
        return None

    # Define the 7 target columns (Adjust keys based on your actual file)
    required_columns = {
        'INSTNM': 'Name',
        'STABBR': 'State',
        'CITY': 'City',
        'SECTOR': 'Sector_Code',
        'HBCU': 'HBCU_Status',
        'LATITUDE': 'Latitude',
        'LONGITUD': 'Longitude'
    }

    # Keep only columns present in the file to avoid key errors
    available_cols = [col for col in required_columns.keys() if col in df.columns]
    df = df[available_cols].copy()

    # Rename columns for readability
    df.rename(columns=required_columns, inplace=True)

    # Drop records missing spatial coordinates
    df.dropna(subset=['Latitude', 'Longitude'], inplace=True)

    # Map raw numeric Sector codes to human-readable strings if needed
    # Example: 1 = Public 4-year, 2 = Private non-profit 4-year, etc.
    sector_mapping = {
        1: "Public, 4-year+", 2: "Private non-profit, 4-year+", 3: "Private for-profit, 4-year+",
        4: "Public, 2-year", 5: "Private non-profit, 2-year", 6: "Private for-profit, 2-year"
    }
    if 'Sector_Code' in df.columns:
        df['Sector'] = df['Sector_Code'].map(sector_mapping).fillna("Other Institutional Sector")

    return df

# ==========================================
# 2. MAP GENERATION FUNCTION
# ==========================================
def build_map(dataframe, output_filename="college_map.html"):
    """
    Generates an interactive HTML map from a filtered DataFrame.
    """
    if dataframe.empty:
        print("⚠️ No matching schools found to plot on a map.")
        return False

    # Start map at a geographic midpoint of the Continental US
    us_center = [39.8283, -98.5795]
    college_map = folium.Map(location=us_center, zoom_start=4, tiles="OpenStreetMap")

    # Using a Marker Cluster to maintain fast performance with thousands of institutions
    marker_cluster = FastMarkerCluster([]).add_to(college_map)

    # Add an individual marker details loop
    for _, row in dataframe.iterrows():
        popup_html = f"""
        <div style="font-family: Arial, sans-serif; font-size: 12px; min-width: 200px;">
            <h4 style="margin: 0 0 5px 0; color: #2C3E50;">{row['Name']}</h4>
            <b>Location:</b> {row['City']}, {row['State']}<br>
            <b>Type:</b> {row.get('Sector', 'N/A')}<br>
            <b>HBCU:</b> {'Yes' if row.get('HBCU_Status') == 1 else 'No'}
        </div>
        """

        folium.Marker(
            location=[row['Latitude'], row['Longitude']],
            popup=folium.Popup(popup_html, max_width=300),
            tooltip=row['Name']
        ).add_to(college_map)

    # Save output HTML
    college_map.save(output_filename)
    print(f"🌐 Success! Map successfully saved to: '{output_filename}'")
    return True

# ==========================================
# 3. INTERACTIVE FILTERING MENU
# ==========================================
def interactive_menu(df):
    """
    Command line interface loop to dynamically filter data based on user choices.
    """
    while True:
        print("\n" + "="*50)
        print("🎓 U.S. COLLEGES & UNIVERSITIES INTERACTIVE MAPPER 🎓")
        print("="*50)
        print("1. Map ALL Institutions")
        print("2. Filter Map by State (e.g., NY, CA, TX)")
        print("3. Map Only HBCU Institutions")
        print("4. Exit Program")

        choice = input("\nEnter choice (1-4): ").strip()

        if choice == '1':
            print(f"\nProcessing all {len(df):,} records...")
            # Display sample of dataframe table
            print(df[['Name', 'City', 'State']].head(10).to_string())
            build_map(df, "all_colleges_map.html")

        elif choice == '2':
            state_input = input("Enter 2-letter state abbreviation: ").strip().upper()
            filtered_df = df[df['State'] == state_input]

            if not filtered_df.empty:
                print(f"\nFound {len(filtered_df):,} institutions in {state_input}:")
                print(filtered_df[['Name', 'City']].head(10).to_string())
                build_map(filtered_df, f"{state_input.lower()}_colleges_map.html")
            else:
                print(f"❌ No records found matching state abbreviation: '{state_input}'")

        elif choice == '3':
            # Check if HBCU column is populated (Typically 1 = Yes)
            if 'HBCU_Status' in df.columns:
                filtered_df = df[df['HBCU_Status'] == 1]
                print(f"\nFound {len(filtered_df)} Historically Black Colleges and Universities:")
                print(filtered_df[['Name', 'City', 'State']].to_string())
                build_map(filtered_df, "hbcu_colleges_map.html")
            else:
                print("❌ HBCU status column not found or formatted incorrectly in source file.")

        elif choice == '4':
            print("\nExiting program. Happy analysis!")
            break
        else:
            print("❌ Invalid entry. Please choose a option between 1 and 4.")

# ==========================================
# MAIN EXECUTION
# ==========================================
if __name__ == "__main__":
    # REPLACE with your path or downloaded IPEDS csv filename
    DATA_FILE = "hd2025.csv"

    # Generate dummy data block if real file doesn't exist yet for seamless testing
    if not os.path.exists(DATA_FILE):
        print(f"ℹ️ '{DATA_FILE}' not spotted. Generating temporary sandbox data for testing...")
        sandbox_data = {
            'INSTNM': ['Columbia University', 'Stanford University', 'Howard University', 'Spelman College', 'Rice University'],
            'STABBR': ['NY', 'CA', 'DC', 'GA', 'TX'],
            'CITY': ['New York', 'Stanford', 'Washington', 'Atlanta', 'Houston'],
            'SECTOR': [1, 2, 1, 2, 1],
            'HBCU': [0, 0, 1, 1, 0],
            'LATITUDE': [40.8075, 37.4275, 38.9226, 33.7464, 29.7174],
            'LONGITUD': [-73.9626, -122.1697, -77.0194, -84.4131, -95.4018]
        }
        pd.DataFrame(sandbox_data).to_csv(DATA_FILE, index=False)

    # Run application
    cleaned_df = load_and_clean_data(DATA_FILE)
    if cleaned_df is not None:
        interactive_menu(cleaned_df)
```
