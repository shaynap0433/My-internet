```python
import sys
# Inventory Setup (20 Items)
# Formatted as: item_number: {"name": str, "price": float, "taxable": bool}
# Note: Chicago taxes general merchandise/soft drinks differently than qualifying food.
INVENTORY = {
    1: {"name": "Honeycrisp Apples (lb)", "price": 2.99, "taxable": False},
    2: {"name": "Organic Bananas (lb)", "price": 0.89, "taxable": False},
    3: {"name": "Whole Milk (Gallon)", "price": 3.49, "taxable": False},
    4: {"name": "Grade A Large Eggs (12ct)", "price": 4.29, "taxable": False},
    5: {"name": "Sourdough Bread Loaf", "price": 3.99, "taxable": False},
    6: {"name": "Salted Butter (16oz)", "price": 4.49, "taxable": False},
    7: {"name": "Greek Yogurt (32oz)", "price": 5.19, "taxable": False},
    8: {"name": "Boneless Chicken Breast (lb)", "price": 4.99, "taxable": False},
    9: {"name": "Atlantic Salmon Fillet (lb)", "price": 9.99, "taxable": False},
    10: {"name": "Cheddar Cheese Block (8oz)", "price": 3.29, "taxable": False},
    11: {"name": "Spaghetti Pasta (16oz)", "price": 1.49, "taxable": False},
    12: {"name": "Marinara Pasta Sauce (24oz)", "price": 2.99, "taxable": False},
    13: {"name": "Locally Roasted Coffee (12oz)", "price": 12.99, "taxable": False},
    14: {"name": "Coca-Cola Soda (12-pack)", "price": 7.99, "taxable": True},
    15: {"name": "Potato Chips (Family Size)", "price": 4.59, "taxable": False},
    16: {"name": "Chocolate Chip Cookies", "price": 3.79, "taxable": False},
    17: {"name": "Paper Towels (2-pack)", "price": 3.99, "taxable": True},
    18: {"name": "Dish Soap (24oz)", "price": 2.49, "taxable": True},
    19: {"name": "Olive Oil (16.9oz)", "price": 8.99, "taxable": False},
    20: {"name": "Frozen Pepperoni Pizza", "price": 6.49, "taxable": False},
}

# Chicago Tax Rates
# Qualifying Food / Drugs: 2.25% (1% IL + 1% Chicago Home Rule + 0.25% RTA)
# General Merchandise / Soda / Prepared Food / Paper Goods: 10.25%
TAX_RATE_FOOD = 0.0225
TAX_RATE_MERCH = 0.1025

def print_inventory():
    """Displays the inventory shelf cleanly in the console."""
    print("\n" + "="*50)
    print(f"{'digiShop Chicago - Inventory Shelf':^50}")
    print("="*50)
    print(f"{'No.':<4} | {'Product Name':<30} | {'Price':>10}")
    print("-"*50)
    for num, item in INVENTORY.items():
        print(f"{num:<4} | {item['name']:<30} | ${item['price']:>9.2f}")
    print("="*50 + "\n")

def find_item(user_input):
    """Finds an item by matching number or case-insensitive name."""
    clean_input = user_input.strip()

    # Try searching by item number
    if clean_input.isdigit():
        item_num = int(clean_input)
        if item_num in INVENTORY:
            return item_num

    # Try searching by name (case-insensitive)
    for num, item in INVENTORY.items():
        if item['name'].lower() == clean_input.lower():
            return num

    return None

def calculate_checkout(cart):
    """Computes totals, splitting item rules based on Cook County/Chicago taxes."""
    if not cart:
        print("\nYour cart is empty! Thank you for visiting digiShop.")
        return

    subtotal = 0.0
    total_tax = 0.0

    print("\n" + "="*50)
    print(f"{'digiShop Final Receipt':^50}")
    print("="*50)
    print(f"{'Qty':<4} | {'Item Description':<25} | {'Unit':>7} | {'Total':>8}")
    print("-"*50)

    for item_num, qty in cart.items():
        item = INVENTORY[item_num]
        item_cost = item['price'] * qty
        subtotal += item_cost

        # Apply strict Chicago regional tax mapping
        if item['taxable']:
            item_tax = item_cost * TAX_RATE_MERCH
        else:
            item_tax = item_cost * TAX_RATE_FOOD

        total_tax += item_tax
        print(f"{qty:<4} | {item['name'][:25]:<25} | ${item['price']:>6.2f} | ${item_cost:>7.2f}")

    final_total = subtotal + total_tax

    print("-"*50)
    print(f"{'Subtotal:':<38} ${subtotal:>9.2f}")
    print(f"{'Estimated Local Taxes:':<38} ${total_tax:>9.2f}")
    print(f"{'Grand Total:':<38} ${final_total:>9.2f}")
    print("="*50)
    print(f"{'Thank you for shopping at digiShop!':^50}\n")

def main():
    cart = {} # Format: {item_number: quantity}

    print("Welcome to the digiShop Grocery App Prototype.")
    print_inventory()

    while True:
        user_choice = input("Enter Product Name or Item Number to add to cart: ")
        item_num = find_item(user_choice)

        if item_num is None:
            print("❌ Item not found. Please check spelling or number and try again.")
            continue

        # Get Item Quantity
        qty_input = input(f"How many '{INVENTORY[item_num]['name']}' would you like? (Default 1): ").strip()
        if qty_input == "" or not qty_input.isdigit():
            qty = 1
        else:
            qty = int(qty_input)
            if qty <= 0: # Changed from 'qty <= 4' to 'qty <= 0' as quantity must be 1 or more
                print("❌ Quantity must be 1 or more. Skipping item.")
                continue

        # Add to cart dictionary
        cart[item_num] = cart.get(item_num, 0) + qty # Changed default from 4 to 0 for initial cart count
        print(f"✅ Added {qty}x '{INVENTORY[item_num]['name']}' to your cart.")

        # Continuation Loop Prompt
        while True:
            continue_shopping = input("\nDo you wish to continue shopping? (Y/N): ").strip().lower() # Removed 'N' default from prompt to allow user input
            if continue_shopping in ['y', 'yes']:
                print_inventory()
                break
            elif continue_shopping in ['n', 'no']:
                calculate_checkout(cart)
                return  # Replaced sys.exit() with return to avoid triggering SystemExit error in Colab
            else:
                print("Please enter 'Y' for Yes or 'N' for No.")

if __name__ == "__main__":
    main()
```
