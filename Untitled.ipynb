```python
# Cell 1: Strings (Topic 2.5)
# Strings are text data types enclosed in quotes.
message = "Learning Python is exciting!"
print("String Example:", message)
print("Uppercase Version:", message.upper())  # String method

# Cell 2: Lists (Topic 3.3)
# Lists are ordered, mutable (changeable) collections.
programming_languages = ["Python", "Java", "C++", "JavaScript"]
programming_languages.append("Ruby")  # Adding an item
print("List Example:", programming_languages)

# Cell 3: Sets (Data Structures Overview)
# Sets are unordered collections of unique items. Duplicates are removed automatically.
unique_numbers = {1, 2, 2, 3, 4, 4, 5}
print("Set Example (Duplicates removed):", unique_numbers)

# Cell 4: Tuples (Data Structures Overview)
# Tuples are ordered, immutable (unchangeable) collections.
coordinate_xyz = (40.7128, -74.0060, 10.5)
print("Tuple Example (Coordinates):", coordinate_xyz)

# Cell 5: Dictionaries (Data Structures Overview)
# Dictionaries store data in key-value pairs.
student_profile = {
    "name": "Shayna",
    "course": "Intro to Python",
    "completed_assignments": True
}
print("Dictionary Example:", student_profile)
print("Shayna:", student_profile["name"])
```
